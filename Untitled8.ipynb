```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\user\Downloads\social_media_dopamine_productivity.csv")
df.head(5)
```

```python
df.info()
```

```python
df.shape
```

```python
df = df.drop(columns=["data_source", "synthetic_flag"])
print("updated Dataset shape:", df.shape)
df.info()
```

```python
bins = [-np.inf, 3, 5, 7.5, np.inf]
labels = ['Good(No addiction)', 'Low Addiction', 'Medium Addiction', 'High Addiction']

df['addiction_factor'] = pd.cut(df['sm_addiction_risk_score'], bins=bins, labels=labels, right=False)
print(df['addiction_factor'].value_counts())
```

```python
trends_data = df.groupby('avg_daily_screen_time_hours')['productivity_self_rating'].mean().reset_index()
plt.figure(figsize=(8, 5))
plt.plot(trends_data['avg_daily_screen_time_hours'], trends_data['productivity_self_rating'], marker='o', color='crimson', linewidth=2, label='Avg Productivity')
plt.title('Productivity Drop off by screen time', fontsize=14, fontweight='bold')
plt.xlabel('avg_daily_screen_time_hours', fontsize=11)
plt.ylabel('productivity_self_rating', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
Most_used_platform = df.groupby('primary_platform')['participant_id'].count().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
Most_used_platform.plot(kind='bar', color='mediumseagreen')
plt.title("Platform Count Distribution")
plt.xlabel('primary_platform')
plt.ylabel('COUNT of participants')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
data = df.groupby('addiction_factor', observed=False)['deep_work_duration_minutes'].mean()
logical_order = ['Good(No addiction)', 'Low Addiction', 'Medium Addiction', 'High Addiction']
data = data.reindex(logical_order)
plt.figure(figsize=(10, 6))
data.plot(kind='bar', color='coral')
plt.title('addiction Factor vs. Focus')
plt.xlabel('addiction_factor', fontsize=11)
plt.ylabel('deep_work_duration_minutes', fontsize=11)
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
platform_counts = df['primary_platform'].value_counts()
print(platform_counts)
```

```python
employment_status = df['employment_status'].value_counts()
plt.figure(figsize=(10, 5))
ax = employment_status.plot(kind='barh', color='skyblue', edgecolor='black')
plt.title('Employement_status among partcipants', fontweight='bold')
plt.xlabel('Count of Participants', fontsize=11)
plt.ylabel('employment_status', fontsize=11)
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
gender_status = df['gender'].value_counts()
plt.figure(figsize=(8, 5))
plt.pie(gender_status, labels = gender_status.index, autopct='%1.1f%%', startangle=140)
plt.title("gender of participants")
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(9, 6))

# Create the scatter plot with a regression trendline
sns.regplot(
    data=df, 
    x='avg_daily_screen_time_hours', 
    y='sm_addiction_risk_score',
    scatter_kws={
        'alpha': 0.6,          # Makes points semi-transparent to reveal clusters/density
        'color': 'teal',       # Marker color
        's': 50                # Marker size
    },
    line_kws={
        'color': 'crimson',    # Trendline color
        'linewidth': 2         # Trendline thickness
    }
)

# Customize titles and labels
plt.title("Screen Time vs. Social Media Addiction Risk", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Average Daily Screen Time (Hours)", fontsize=11)
plt.ylabel("Social Media Addiction Risk Score (1-10)", fontsize=11)

# Add a clean grid and layout adjustment
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Display the plot
plt.show()
```
